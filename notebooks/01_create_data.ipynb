```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

np.random.seed(42)
random.seed(42)
n = 500

banks = ['SBI', 'HDFC', 'ICICI', 'Axis', 'Kotak']
categories = ['UPI', 'NEFT', 'IMPS', 'RTGS', 'ATM']
merchants = ['Zomato', 'Swiggy', 'Amazon', 'Flipkart',
             'IRCTC', 'PhonePe', 'Paytm', 'BigBasket',
             'Ola', 'Uber', 'MakeMyTrip', 'BookMyShow']

# Bank Ledger
bank_ledger = pd.DataFrame({
    'txn_id': [f'TXN{str(i).zfill(5)}' for i in range(n)],
    'date': [datetime(2024,1,1) + timedelta(days=random.randint(0,180)) for _ in range(n)],
    'amount_inr': np.round(np.random.uniform(50, 50000, n), 2),
    'type': np.random.choice(categories, n),
    'bank': np.random.choice(banks, n),
    'merchant': np.random.choice(merchants, n),
    'status': np.random.choice(['SUCCESS','FAILED','PENDING'], n, p=[0.85, 0.08, 0.07])
})

# UPI Statement with 10% discrepancies injected
upi_statement = bank_ledger.copy()
noise_idx = np.random.choice(n, int(n * 0.10), replace=False)
upi_statement.loc[noise_idx, 'amount_inr'] = (
    upi_statement.loc[noise_idx, 'amount_inr'] *
    np.random.uniform(0.90, 1.10, len(noise_idx))
).round(2)
upi_statement.loc[noise_idx, 'status'] = 'DISCREPANCY'

bank_ledger.to_csv('../data/bank_ledger.csv', index=False)
upi_statement.to_csv('../data/upi_transactions.csv', index=False)

print(f"✅ bank_ledger.csv — {len(bank_ledger)} rows")
print(f"✅ upi_transactions.csv — {len(upi_statement)} rows")
print(f"🔴 Discrepancies injected: {len(noise_idx)}")
print("\nSample data:")
print(bank_ledger.head(3))
```
